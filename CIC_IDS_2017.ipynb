```python
import os
import pandas as pd
import numpy as np
# loads the env
from dotenv import load_dotenv
import kagglehub

# 1. Load the env file
load_dotenv()

# 2. Assign credentials (FIXED: KAGGLE_KEY now maps to the correct variable)
os.environ['KAGGLE_USERNAME'] = os.getenv('KAGGLE_USERNAME', '')
os.environ['KAGGLE_KEY'] = os.getenv('KAGGLE_KEY', '')

import kaggle 

# 3. Attempting Authentication
try:
    kaggle.api.authenticate()
    print("Successfully obtained env/creds :)")
except Exception as e:
    print(f"Authentication unfortunately failed :( : {e}")

path = kagglehub.dataset_download("chethuhn/network-intrusion-dataset")
print(" Path to dataset files:", path)
```

```python
files = os.listdir(path)
print(f"Dataset has {len(files)} files:\n" + "\n".join(files))
```

```python
dfs = []
for file in files:
    df = pd.read_csv(os.path.join(path, file))
    df.columns = df.columns.str.strip()
    dfs.append(df)
    print(df.shape)

data = pd.concat(dfs, ignore_index=True)

columns_drop = [
    'Fwd Header Length.1',
    'Avg Fwd Segment Size',
    'Subflow fwd bytes',
    'Subflow fwd packets',
    'Subflow bwd Packets',
    'Subflow bwd bytes',
    'Packet Length Variance' 
]

data = data.drop(columns=columns_drop, errors='ignore')

for col in data.columns:
    print(col)
```
